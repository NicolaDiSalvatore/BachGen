 # BachGen Training


This notebook can be used on Colab to train BachGen. Here DagsHub is used to track experiments with Mlflow. The secrets to run the notebook are:
- GITHUB_TOKEN: the token associated with the BachGen repository on GitHub
- DAGSHUB_USERNAME: the username of the DagsHub account
- DAGSHUB_TOKEN: a token generated from the DagsHub account
- DAGSHUB_REPO_NAME: the name of the DagsHub repository associated with the BachGen repository on GitHub

```python
from google.colab import userdata
import os
token = userdata.get('GITHUB_TOKEN')
os.environ["MLFLOW_TRACKING_URI"] = "https://dagshub.com/{}/{}.mlflow".format(
    userdata.get("DAGSHUB_USERNAME"),
    userdata.get("DAGSHUB_REPO_NAME")
)
print("✅ URI set:", os.environ["MLFLOW_TRACKING_URI"])
!git clone https://{token}@github.com/NicolaDiSalvatore/BachGen.git
```
